### a. Write a Python script to perform PCA (Principal Component Analysis) on a dataset. Assume the dataset has numerical features only.

```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
data = {
    'user_id': [23456, 54213, 12345, 67890, 23457],
    'trxn_amt': [10, 20, 15, 25, 12],
    'acc_bal': [100, 200, 150, 300, 120],
    'f1': [1.0, 2.0, 1.5, 2.5, 1.2],
    'f2': [3.0, 4.0, 3.5, 4.5, 3.2]
}
df = pd.DataFrame(data)
```

```python
def pca(df):
    features = df.select_dtypes(include=[float, int])
    features_standardized = StandardScaler().fit_transform(features)

    pca = PCA(n_components=2)
    principal_components = pca.fit_transform(features_standardized)

    pca_df = pd.DataFrame(data=principal_components, columns=['PC 1', 'PC 2'])
    return pca_df

pca_result = pca(df)
print("PCA:")
print(pca_result)
```

### c. Write a Python function that creates a new feature 'transaction_ratio' in a DataFrame. This feature is the ratio of 'transaction_amount' to 'account_balance'

```python
def create_transaction_ratio(df):
    df['transaction_ratio'] = df['transaction_amount'] / df['account_balance']
    return df

df_with_ratio = create_transaction_ratio(df)
print("\nDataFrame with Transaction Ratio:")
print(df_with_ratio)
```
