### a. Write a Python function that takes a DataFrame containing transaction data and aggregates the total transaction amount per user. The DataFrame has columns 'user_id', 'item_name' and 'transaction_amount'.

```python
import pandas as pd
```

```python
df = pd.read_csv('sample_transaction.csv', sep = ', ')
df.head()
df.shape
```

```python
def aggregator(df):
    df_grouped = df
    sum = df_grouped.groupby('user_id')['transaction_amount'].sum()
    sum.name = 'total_purchase'
    print(sum)

aggregator(df)
```
