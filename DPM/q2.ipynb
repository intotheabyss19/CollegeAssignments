### (b) Replace missing values (encoded as '?') with NaNs in the breast cancer dataset.

```python
def replace_missing_values(df):
    for column in df:
        for row in df[column]:
            if row == '?':
                print("Missing Value")
                row = nan
                
replace_missing_values(X)
```
