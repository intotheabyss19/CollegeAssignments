Enter temperature and humidity values. The program checks them against the safe limits and reports the
greenhouse status, the LED that would light up, and the action to take (same logic as the Tinkercad Arduino circuit).

## Alert thresholds
| Parameter | Safe range | If too low | If too high |
|---|---|---|---|
| Temperature | 15 – 35 °C | Heater ON | Exhaust fan ON |
| Humidity | 40 – 85 %RH | Misting ON | Ventilation ON |

```python
TEMP_LOW, TEMP_HIGH = 15.0, 35.0   # °C
HUM_LOW, HUM_HIGH = 40.0, 85.0     # %RH
```

## Alert logic

```python
def check_greenhouse(temp, hum):
    alerts = []

    temp_alert = temp < TEMP_LOW or temp > TEMP_HIGH
    if temp > TEMP_HIGH:
        alerts.append("HIGH TEMPERATURE -> Exhaust fan ON")
    elif temp < TEMP_LOW:
        alerts.append("LOW TEMPERATURE -> Heater ON")

    hum_alert = hum < HUM_LOW or hum > HUM_HIGH
    if hum < HUM_LOW:
        alerts.append("LOW HUMIDITY -> Misting ON")
    elif hum > HUM_HIGH:
        alerts.append("HIGH HUMIDITY -> Ventilation ON")

    if len(alerts) == 0:
        status = "NORMAL"
    elif len(alerts) == 1:
        status = "WARNING"
    else:
        status = "CRITICAL"

    print(f"Temp: {temp:.1f} C | Humidity: {hum:.1f} % | Status: {status}")
    for a in alerts:
        print("  ALERT:", a)
    print("  Green LED :", "ON" if status == "NORMAL" else "OFF")
    print("  Red LED   :", "ON" if temp_alert else "OFF")
    print("  Yellow LED:", "ON" if hum_alert else "OFF")
```

## Enter readings
Type a temperature and humidity for each reading; type `q` to stop.

```python
while True:
    t = input("Temperature (C) or q to quit: ")
    if t.strip().lower() == "q":
        break
    h = input("Humidity (%RH): ")
    try:
        temp, hum = float(t), float(h)
    except ValueError:
        print("Please enter numbers only.\n")
        continue
    if not 0 <= hum <= 100:
        print("Humidity must be between 0 and 100.\n")
        continue
    check_greenhouse(temp, hum)
    print()
```
