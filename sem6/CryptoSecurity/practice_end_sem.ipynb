## Newton raphson for sq root

```python
def nr(n, prec = 2):
    x = n
    while abs(n - x*x) > 10**-prec:
        x = (x + n/x) / 2
    return x

n = 16
print(nr(16, 9))
```

## Extended Euclid's

```python
def ext(a, b):
    if b == 0:
        return a, 1, 0
    g, s1, t1 = ext(b, a%b)
    s = t1
    t = s1 - a//b * s
    return g, s, t

a = 273
b = 98
g, s, t = ext(a, b)
print(g, s, t)
```

## LCM using PF

```python
def pf(n):
    factors = []
    for i in range(2, int(n**0.5) + 1, 1):
        while n % i == 0:
            factors.append(i)
            n //= i
    if n != 1:
        factors.append(n)
    return factors
    

from collections import Counter

def lcm(a, b):
    a_pf = Counter(pf(a))
    b_pf = Counter(pf(b))
    s = set(a_pf.keys() | b_pf.keys())
    res = 1
    for p in s:
        maxi = max(a_pf.get(p, 0), b_pf.get(p, 0))
        res *= p ** maxi
    return res
print(lcm(20, 60))
```

## Ceaser Cipher

```python
def caesar(text, shift=3, decrypt=False):
    if decrypt: shift *= -1
    res = []
    for t in text:
        if not t.isalpha(): 
            res.append(t)
        else:
            val = (ord(t) - 65) % 26
            r = (val + shift) % 26 + 65
            res.append(chr(r))
    return ''.join(r for r in res)

cipher = caesar(text)
print(cipher)
print(caesar(cipher, decrypt=True))
```

## Vigenere Cipher

```python
def vigenere(text, key, decrypt=False):
    if decrypt: d=-1
    else: d=1
    res = []
    keyl = list(key.replace(' ', ''))
    i = 0
    for t in text:
        if not t.isalpha():
            res.append(t)
        else:
            shift = ord(keyl[i%len(key)]) - 65
            val = (ord(t) - 65) % 26
            r = (val + shift*d) % 26 + 65
            i += 1
            res.append(chr(r))
    return ''.join(r for r in res)

# text = 'SATYA'
cipher = vigenere(text, key)
print(cipher)
print(vigenere(cipher, key, decrypt=True))
```

## Hill Cipher

```python
def hill(text, key, decrypt=False):
    # text = text.replace(' ', '')
    if decrypt:
        inv = pow(key[0][0]*key[1][1] - key[0][1]*key[1][0], -1, 26)
        key = [
            [key[1][1]*inv%26, -key[0][1]*inv%26], 
            [-key[1][0]*inv%26, key[0][0]*inv%26]
            ]

    t = [ord(c) - 65 for c in text if c.isalpha()]

    if len(t)%2:
        t.append(23)
    print(t)
    res = []
    for i in range(0, len(t), 2):
        c_block = [key[0][0]*t[i] + key[0][1]*t[i+1], key[1][0]*t[i] + key[1][1]*t[i+1]]
        res.append(chr(c_block[0]%26 + 65))
        res.append(chr(c_block[1]%26 + 65))

    return ''.join(r for r in res)


key = [[3, 3], [2, 5]]   
text = "NATIONAL SIKKIM"
cipher = hill(text, key)
print(cipher)
print(hill(cipher, key, decrypt=True))
```

```python
def rail(text, rails, decrypt=False):
    seq = list(range(rails)) + list(range(rails-2, 0, -1))

    if not decrypt: #encryption
        res = [""] * rails
        for i, c in enumerate(text):
            res[seq[i % len(seq)]] += c
        return ''.join(res)
    
    #decryption
    counts = [0] * rails

    for i in range(len(text)):
        counts[seq[i % len(seq)]] += 1
    print('counts', counts)

    r_lst, idx = [], 0
    for c in counts:
        r_lst.append(list(text[idx : idx + c]))
        idx += c
    print(r_lst)

    return "".join(r_lst[seq[i % len(seq)]].pop(0) for i in range(len(text)))


text = 'NATIONAL SIKKIM'
cipher = rail(text, 4)
print(cipher)
print(rail(cipher, 4, decrypt=True))
```
