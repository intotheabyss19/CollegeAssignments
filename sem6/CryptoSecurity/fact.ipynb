```python
def factorization(n):
    factors = []
    for i in range(2, n//2 + 1):
        if n < 1:
            breakd
        while n % i == 0:
            factors.append(i)
            n /= i
    if n > 1:
        factors.append(n)
    return factors

from collections import Counter

def prime_fac(a, b):
    factors_a = factorization(a)
    factors_b = factorization(b)
    a = Counter(factors_a)
    b = Counter(factors_b)
    factors = set(a.keys() | b.keys())
    lcm = 1
    for p in factors:
        maxi = max(a.get(p, 0), b.get(p, 0))
        lcm *= p ** maxi

```
