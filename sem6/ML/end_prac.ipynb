### (a) Write a program in Python to check whether a given number is Palindrome or not.

```python
n = 10001
num = str(n)
for i in range(len(num)):
    if num[i] != num[-i-1]: 
        print(False)
        break
    continue
else: print(True)
```

### (b) Write a function in Python that returns factorial of a given number.

```python
def fact(n):
    if n <= 1: return 1
    return n * fact(n-1)
print(fact(6))
```
