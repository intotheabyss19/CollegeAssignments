### Function Minimizing using Gradient Descent
            f(x,y,z)=x^2+2y^2+5z^2

```python
def f(x, y, z):
    return x**2 + 2*y**2 + 5*z**2

def gradient_descent_minimize(x, y, z, lr=0.1):
    print(f"initial_values: x={x}, y={y}, z={z} | f(x,y,z)={f(x,y,z):.6f}\n")
    for i in range (1, 20):
        grad_x = 2*x
        grad_y = 4*y
        grad_z = 10*z

        x = x - lr * grad_x
        y = y - lr * grad_y
        z = z - lr * grad_z

        f_new = f(x,y,z)
        print(f"X = {x:.4f} , Y = {y:.4f}, Z = {z:.4f} | f(x,y,z) = {f_new:.4f} ")

gradient_descent_minimize(1,1,1,0.1)
```

Find out the standard deviation of N=10 random numbers which are randomly generated between the range of 0 and 1.

```python
import random

nums = [random.uniform(0,1) for _ in range(10)]

nums_mean = sum(nums)/len(nums)

variance_sum = 0

for x in nums:
    variance_sum += (x - nums_mean)**2

std = (variance_sum / len(nums)) ** 0.5

print("Mean: =",nums_mean)
print("Standard Deviation: ",std)
```

```python
import numpy as np

def forward_algorithm(pi, A, B, observations, state_labels, obs_labels):
    pi = np.array(pi)
    A  = np.array(A)
    B  = np.array(B)
    T  = len(observations)
    N  = len(pi)

    alpha = np.zeros((T, N))

    # Initialization
    o1 = observations[0]
    alpha[0] = pi * B[:, o1]

    # Induction
    for t in range(1, T):
        ot = observations[t]
        for j in range(N):
            summation = sum(alpha[t-1, i] * A[i, j] for i in range(N))
            alpha[t, j] = summation * B[j, ot]

    # Termination
    prob = alpha[T-1].sum()

    return prob


state_labels = ["S", "C", "R"]
obs_labels = ["W", "Sh", "Cl"]
pi = [0.5, 0.3, 0.2]
A = [
    [0.6, 0.3, 0.1],
    [0.4, 0.4, 0.2],
    [0.3, 0.3, 0.4],
]
B = [
    [0.6, 0.3, 0.1],
    [0.3, 0.4, 0.3],
    [0.1, 0.3, 0.6],
]
observations = [1, 1, 0]

prob = forward_algorithm(
    pi, A, B, observations,
    state_labels,
    obs_labels
)
print(prob)
```
