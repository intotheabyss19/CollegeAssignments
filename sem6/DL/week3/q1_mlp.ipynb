# Q1 – MLP Binary Classification
### CS16204 | Adult, Liver, Titanic Datasets

```python
import pandas as pd
import numpy as np
from sklearn.neural_network import MLPClassifier as MLP
from sklearn.model_selection import train_test_split as tts
from sklearn.preprocessing import LabelEncoder as LE, StandardScaler as SS
from sklearn.metrics import accuracy_score as acc, classification_report as cr
```

## Helper: Train & Evaluate

```python
def run(X, y, name):
    sc = SS()
    Xt, Xv, yt, yv = tts(X, y, test_size=0.3, random_state=42)
    Xt = sc.fit_transform(Xt)
    Xv = sc.transform(Xv)
    m = MLP(hidden_layer_sizes=(64,32), max_iter=300, random_state=42)
    m.fit(Xt, yt)
    yp = m.predict(Xv)
    print(f'\n=== {name} ===')
    print(f'Accuracy: {acc(yv, yp):.4f}')
    print(cr(yv, yp))
```

## 1a. Adult Census Income
> Download: https://www.kaggle.com/datasets/uciml/adult-census-income  
> File: `adult.csv`

```python
df = pd.read_csv('adult.csv')
df.replace('?', np.nan, inplace=True)
df.dropna(inplace=True)

le = LE()
for c in df.select_dtypes('object').columns:
    df[c] = le.fit_transform(df[c])

X = df.drop('income', axis=1).values
y = df['income'].values

run(X, y, 'Adult Census Income')
```

## 1b. Indian Liver Patient Records
> Download: https://www.kaggle.com/datasets/uciml/indian-liver-patient-records  
> File: `indian_liver_patient.csv`

```python
df = pd.read_csv('indian_liver_patient.csv')
df.dropna(inplace=True)

le = LE()
df['Gender'] = le.fit_transform(df['Gender'])

X = df.drop('Dataset', axis=1).values
y = df['Dataset'].values   # 1=needs diagnosis, 2=healthy

run(X, y, 'Indian Liver Patient')
```

## 1c. Titanic Survival
> Download: https://www.kaggle.com/c/titanic/data  
> File: `train.csv`

```python
# See which columns have NaN
print(df.isnull().sum())

# Fill ALL remaining NaNs at once
for col in df.columns:
    if df[col].dtype == 'object':
        df[col].fillna(df[col].mode()[0], inplace=True)
    else:
        df[col].fillna(df[col].median(), inplace=True)

# Encode any remaining object columns
df = pd.get_dummies(df, drop_first=True)

print(df.isnull().sum())  # Should be all 0s now
run(X, y, 'Titanic Survival')
```
