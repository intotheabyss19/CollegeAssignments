```python
import numpy as np
import pandas as pd
```

```python
def mp_neuron(x, th):
    if np.sum(x) > th:
        return 1
    else:
        return 0
```

```python
np.random.seed(0)

age = np.random.randint(20, 60, 500)
sal = np.random.randint(1, 10, 500)
fam = np.random.randint(1, 6, 500)

df = pd.DataFrame({
    'age': age,
    'sal': sal,
    'fam': fam
})

df['y'] = df['age'] + df['sal'] + df['fam']
df['y'] = df['y'].apply(lambda x: 1 if x > 50 else 0)
```

```python
# Split
sp = int(0.7 * len(df))

df_tr = df.iloc[:sp]
df_te = df.iloc[sp:]
```

```python
th = df_tr[['age','sal','fam']].sum(axis=1).mean()
```

```python
c = 0

for i in range(len(df_te)):
    x = df_te.iloc[i][['age','sal','fam']]
    y_p = mp_neuron(x, th)
    
    if y_p == df_te.iloc[i]['y']:
        c += 1

acc = c / len(df_te)
print("Test Accuracy:", acc)
```
