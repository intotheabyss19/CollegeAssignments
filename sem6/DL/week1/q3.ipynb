```python
import numpy as np
import pandas as pd
```

```python
def perceptron(x, w, b):
    return 1 if (np.dot(x, w) + b) >= 0 else 0
```

```python
np.random.seed(1)
n = 500

age = np.random.randint(18, 70, n)
sal = np.random.randint(30000, 150000, n)
fam = np.random.randint(1, 6, n)

buy_val = (age/70)*0.3 + (sal/150000)*0.5 + (fam/6)*0.2
buy = (buy_val > 0.7).astype(int)

df = pd.DataFrame({
    'age': age,
    'sal': sal,
    'fam': fam,
    'y': buy
})
print(df)
# normalization
df[['age','sal','fam']] = (df[['age','sal','fam']] - 
                            df[['age','sal','fam']].min()) / (
                            df[['age','sal','fam']].max() - 
                            df[['age','sal','fam']].min())
```

```python
sp = int(0.7 * len(df))
tr = df.iloc[:sp]
te = df.iloc[sp:]

X_tr = tr[['age','sal','fam']].values
Y_tr = tr['y'].values

X_te = te[['age','sal','fam']].values
Y_te = te['y'].values
```

```python
# training
w = np.zeros(3)
b = 0
lr = 0.01
epochs = 100

for _ in range(epochs):
    for x, y in zip(X_tr, Y_tr):
        y_hat = perceptron(x, w, b)
        err = y - y_hat
        w = w + lr * err * x
        b = b + lr * err

# testing
c = 0
for x, y in zip(X_te, Y_te):
    if perceptron(x, w, b) == y:
        c += 1

acc = c / len(Y_te)
print("Test Accuracy:", acc)
```
