```python
import numpy as np
import pandas as pd
```

```python
and_data = {
    'x1': [0, 0, 1, 1],
    'x2': [0, 1, 0, 1],
    'y':  [0, 0, 0, 1]
}

df_and = pd.DataFrame(and_data)
```

```python
or_data = {
    'x1': [0, 0, 1, 1],
    'x2': [0, 1, 0, 1],
    'y':  [0, 1, 1, 1]
}

df_or = pd.DataFrame(or_data)
```

```python
class Perceptron:

    def __init__(self, w, b):
        self.w = w      # weights
        self.b = b      # bias (threshold)

    def step(self, s):
        return 1 if s >= self.b else 0

    def predict(self, x):
        s = np.dot(x, self.w)
        return self.step(s)
```

```python
w_and = np.array([1, 1])
b_and = 2

p_and = Perceptron(w_and, b_and)

print("AND Gate Output")
for i in range(len(df_and)):
    x = df_and.iloc[i][['x1','x2']]
    y_pred = p_and.predict(x)
    print(x.values, "->", y_pred)
```

```python
w_or = np.array([1, 1])
b_or = 1

p_or = Perceptron(w_or, b_or)

print("OR Gate Output")
for i in range(len(df_or)):
    x = df_or.iloc[i][['x1','x2']]
    y_pred = p_or.predict(x)
    print(x.values, "->", y_pred)
```
