### 5. Manually adjust the weights and bias of a perceptron to model the XOR logic gate. Discuss why a single-layer perceptron cannot solve the XOR problem effectively. Implement the perceptron learning rule. Use this rule to train a perceptron on a simple dataset (eg. a linearly separable dataset). Plot the decision boundary before and after training.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

Y = np.array([0, 1, 1, 1])
```

```python
class Perceptron:
    def __init__(self, lr=1):
        self.w = np.zeros(2)
        self.b = 0
        self.lr = lr

    def predict(self, x):
        return 1 if (np.dot(x, self.w) + self.b) >= 0 else 0

    def fit(self, X, Y, epochs):
        for _ in range(epochs):
            for x, y in zip(X, Y):
                y_hat = self.predict(x)
                err = y - y_hat
                self.w = self.w + self.lr * err * x
                self.b = self.b + self.lr * err
```

```python
def plot_db(w, b, X, Y, title):
    for i in range(len(Y)):
        if Y[i] == 0:
            plt.scatter(X[i][0], X[i][1], c='red')
        else:
            plt.scatter(X[i][0], X[i][1], c='blue')

    x1 = np.linspace(-0.5, 1.5, 2)
    x2 = -(w[0]*x1 + b) / w[1]
    plt.plot(x1, x2)
    plt.title(title)
    plt.show()
```

```python
# Before Training
p = Perceptron()
plot_db(p.w, p.b, X, Y, "Before Training")
# Train perceptron
p.fit(X, Y, epochs=10)
# After training
plot_db(p.w, p.b, X, Y, "After Training")
```
