### Using a dataset with features on different scales (e.g., "salary" in thousands and "age" in tens), compare the training time of a Sigmoid Neuron on raw data versus normalized data (Min-Max scaling).

```python
import pandas as pd
import numpy as np

# Set seed and generate data
np.random.seed(1)
n = 500
age = np.random.randint(18, 70, n)
sal = np.random.randint(30000, 150000, n)
fam = np.random.randint(1, 6, n)

buy_val = (age/70)*0.3 + (sal/150000)*0.5 + (fam/6)*0.2
buy = (buy_val > 0.7).astype(int)

df = pd.DataFrame({
    'age': age,
    'sal': sal,
    'fam': fam,
    'y': buy
})

# df[['age','sal','fam']] = (df[['age','sal','fam']] - 
#                             df[['age','sal','fam']].min()) / (
#                             df[['age','sal','fam']].max() - 
#                             df[['age','sal','fam']].min())

X = df.iloc[:, :-1].values
Y = df.iloc[:, -1].values
```

```python
class SigmoidNeuron:
    def __init__(self, input_size, eta=1.0):
        self.w = np.zeros(input_size) 
        self.b = 0.0
        self.eta = eta

    def f(self, x):
        weighted_sum = np.dot(x, self.w) + self.b
        return 1.0 / (1.0 + np.exp(-weighted_sum))

    def err(self, X, Y):
        err = 0.0
        for x, y in zip(X, Y):
            fx = self.f(x)
            err += 0.5 * (fx - y) ** 2
        return err
        
    def grad_b(self, x, y):
        fx = self.f(x)
        return (fx - y) * (1 - fx) * fx

    def grad_w(self, x, y):
        fx = self.f(x)
        return (fx - y) * (1 - fx) * fx * x
        
    def do_gradient_descent(self, X, Y):
        max_epochs = 1000
        for i in range(max_epochs):
            dw = np.zeros(self.w.shape) 
            db = 0.0
            
            for x, y in zip(X, Y):
                dw += self.grad_w(x, y)
                db += self.grad_b(x, y)
                
            self.w -= self.eta * dw
            self.b -= self.eta * db
            
        print(f"Training Complete.\nFinal Weights: {self.w}\nFinal Bias: {self.b}\n")

neuron = SigmoidNeuron(input_size=X.shape[1], eta=0.1)
neuron.do_gradient_descent(X, Y)

print("Comparing True Labels to Soft Probabilities")
print(f"{'True Label':<15} | {'Predicted Probability':<25} | {'Interpretation'}")
print("-" * 70)

for i in range(15):
    x_sample = X[i]
    y_actual = Y[i]
    prob = neuron.f(x_sample)
    interpretation = "Likely to Buy" if prob >= 0.5 else "Unlikely to Buy"
    print(f"{y_actual:<15} | {prob:<25.4f} | {interpretation}")
```
