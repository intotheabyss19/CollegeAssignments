```python
import numpy as np

class SigmoidNeuron:
    def __init__(self, input_size, eta=0.1):
        self.w = np.zeros(input_size)
        self.b = 0.0
        self.eta = eta

    def sigmoid(self, z):
        return 1.0 / (1.0 + np.exp(-z))

    def f(self, x):
        return self.sigmoid(np.dot(x, self.w) + self.b)

    def error(self, X, Y):
        err = 0.0
        for x, y in zip(X, Y):
            fx = self.f(x)
            err += 0.5 * (fx - y) ** 2
        return err

    def grad_w(self, x, y):
        fx = self.f(x)
        return (fx - y) * fx * (1 - fx) * x

    def grad_b(self, x, y):
        fx = self.f(x)
        return (fx - y) * fx * (1 - fx)

    def train(self, X, Y, epochs=10000):
        for _ in range(epochs):
            dw = np.zeros_like(self.w)
            db = 0.0
            for x, y in zip(X, Y):
                dw += self.grad_w(x, y)
                db += self.grad_b(x, y)
            self.w -= self.eta * dw
            self.b -= self.eta * db


# XOR dataset
X_xor = np.array([[0, 0],
                  [0, 1],
                  [1, 0],
                  [1, 1]])
Y_xor = np.array([0, 1, 1, 0])

# Train neuron
neuron_xor = SigmoidNeuron(input_size=2, eta=0.1)
print("Training on XOR problem...")
neuron_xor.train(X_xor, Y_xor, epochs=10000)

# Predictions and accuracy
correct = 0
print("\nPredictions on XOR:")
for x, y in zip(X_xor, Y_xor):
    prob = neuron_xor.f(x)
    pred = 1 if prob >= 0.5 else 0
    correct += (pred == y)
    print(f"Input: {x}, True: {y}, Prob: {prob:.4f}, Pred: {pred}")

accuracy = correct / len(Y_xor)
print(f"\nFinal Accuracy on XOR: {accuracy * 100:.2f}%")
print(f"Final Error: {neuron_xor.error(X_xor, Y_xor):.4f}")
```
