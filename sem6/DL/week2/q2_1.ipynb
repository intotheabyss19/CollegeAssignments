### Using a dataset with features on different scales (e.g., "salary" in thousands and "age" in tens), compare the training time of a Sigmoid Neuron on raw data versus normalized data (Min-Max scaling).

```python
import pandas as pd
import numpy as np
import time
import warnings

np.random.seed(42)
n = 1000
age = np.random.randint(18, 70, n)
sal = np.random.randint(30000, 150000, n)
fam = np.random.randint(1, 6, n)

buy_val = (age/70)*0.3 + (sal/150000)*0.5 + (fam/6)*0.2
Y = (buy_val > 0.7).astype(int)

X_raw = np.column_stack((age, sal, fam))

X_norm = (X_raw - X_raw.min(axis=0)) / (X_raw.max(axis=0) - X_raw.min(axis=0))
```

```python
class SigmoidNeuron:
    def __init__(self, input_size, eta=0.01):
        self.w = np.zeros(input_size)
        self.b = 0.0
        self.eta = eta

    def f(self, X):
        return 1.0 / (1.0 + np.exp(-np.clip(np.dot(X, self.w) + self.b, -700, 700)))

    def train(self, X, Y, epochs=1000):
        start_time = time.time()
        
        for _ in range(epochs):
            fx = self.f(X)
            
            error = fx - Y
            dw = np.dot(X.T, error * fx * (1 - fx))
            db = np.sum(error * fx * (1 - fx))
            
            self.w -= self.eta * dw
            self.b -= self.eta * db
            
        end_time = time.time()
        final_loss = 0.5 * np.sum((self.f(X) - Y) ** 2)
        
        return end_time - start_time, final_loss

print("Training on Raw Data")
neuron_raw = SigmoidNeuron(input_size=3, eta=0.01)
time_raw, loss_raw = neuron_raw.train(X_raw, Y, epochs=1000)
print(f"Time Taken: {time_raw:.4f} seconds")
print(f"Final Loss: {loss_raw:.4f} (Notice how high this is!)")
print(f"Weights: {neuron_raw.w}\n")

print("Training on Normalized Data")
neuron_norm = SigmoidNeuron(input_size=3, eta=0.01)
time_norm, loss_norm = neuron_norm.train(X_norm, Y, epochs=1000)
print(f"Time Taken: {time_norm:.4f} seconds")
print(f"Final Loss: {loss_norm:.4f} (Successfully Learned!)")
print(f"Weights: {neuron_norm.w}")
```
