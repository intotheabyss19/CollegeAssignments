# Problem 1: Image Representation and Data Handling

#### 1. Library Comparison: Write a script to read a color image using OpenCV, Matplotlib, and Scikit-Image. Compare their internal storage (BGR vs. RGB) and data types (uint8 vs. float).
#### 2. Manual Channel Manipulation: Extract the Red, Green, and Blue channels. Create a new ”swapped” image where the Red channel is replaced by the Blue channel and visualize the result

```python
import cv2
from matplotlib import pyplot as plt
from matplotlib import image as img
```

Using OpenCV & Manual color channel Manipulation

```python
img_cv_bgr = cv2.imread("mountains.jpg", cv2.IMREAD_COLOR_BGR)
img_cv_rgb = img_cv_bgr[:, :, [2, 1, 0]]
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.imshow(img_cv_bgr)
plt.title('BGR')
plt.subplot(1, 2, 2)
plt.title('RGB')
plt.imshow(img_cv_rgb)
plt.show()
plt.figure(figsize=(10, 6))

plt.subplot(1, 2, 1)
plt.imshow(img_cv_bgr)
plt.show()

print(img_cv_bgr.size/img_cv_rgb.size)
print(type(img_cv_bgr[0,0,0]))
print(img_cv_rgb.size)
print(img_cv_bgr.size/img_cv_bgr.astype(float).size)
print(img_cv_bgr.astype(float).size)
print(type(img_cv_bgr[0,0,0].astype(float)))
```
