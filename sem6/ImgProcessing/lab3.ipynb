### Sampling and Quantization

```python
import cv2, numpy as np
import matplotlib.pyplot as plt

im = cv2.imread('mountains.jpg', 0)

s = 4
sm = im[::s, ::s]

L = 8
q = np.floor(im/(256/L))*(256/L)
q = q.astype(np.uint8)

plt.figure(figsize=(10,4))

plt.subplot(121)
plt.imshow(sm, cmap='gray')
plt.title('Sampled')
plt.axis('off')

plt.subplot(122)
plt.imshow(q, cmap='gray')
plt.title('Quantized')
plt.axis('off')

plt.show()
```

## Histogram Analysis

```python
import cv2
import matplotlib.pyplot as plt

im = cv2.imread('mountains.jpg', 0)

L = 4
q = (im//(256//L))*(256//L)

plt.figure(figsize=(10,4))

plt.subplot(121)
plt.hist(im.ravel(), 256, [0,256])
plt.title('Original')

plt.subplot(122)
plt.hist(q.ravel(), 256, [0,256])
plt.title('Quantized')

plt.show()

print(im.mean())
print(q.mean())
```

## 1D DFT

```python
import cv2, numpy as np
import matplotlib.pyplot as plt

im = cv2.imread('mountains.jpg', 0)

fr = np.fft.fft(im, axis=1)
fc = np.fft.fft(im, axis=0)

plt.figure(figsize=(10,4))

plt.subplot(121)
plt.imshow(np.log(1+np.abs(fr)), cmap='gray')
plt.title('Row FFT')

plt.subplot(122)
plt.imshow(np.log(1+np.abs(fc)), cmap='gray')
plt.title('Col FFT')

plt.show()
```

## 2D FFT Spectrum

```python
im = cv2.imread('mountains.jpg', 0)

f = np.fft.fft2(im)
fs = np.fft.fftshift(f)

m = np.log(1+np.abs(fs))

plt.imshow(m, cmap='gray')
plt.title('2D FFT')
plt.axis('off')
plt.show()
```

## 2D DCT

```python
im = cv2.imread('mountains.jpg', 0)

d = cv2.dct(np.float32(im))

s = np.log(abs(d) + 1)

plt.imshow(s)
plt.title('DCT')
plt.show()
```
