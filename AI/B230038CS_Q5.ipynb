# 5. Write a program to find an optimal path for graph traversal using uniform cost search and interative deepening depth first search strategies.

```python
import heapq

G = {
    'A': [('B', 3), ('C', 5), ('D', 4)],
    'B': [('E', 2), ('F', 4), ('G', 7), ('H', 4)],
    'C': [('F', 3), ('H', 2), ('I', 4), ('O', 7)],
    'D': [('J', 9)],
    'E': [('K', 2)],
    'F': [('K', 1)],
    'G': [],
    'H': [('L', 9), ('M', 7)],
    'I': [('M', 3), ('N', 2)],
    'J': [('O', 3), ('P', 1)],
    'L': [],
    'K': [],
    'M': [],
    'N': [],
    'O': [],
    'P': []
}
```

#### Uniform Cost Search

```python
def reconstruct_path(visited, start, goal):
    path = []
    current = goal
    while current is not None:
        path.append(current)
        current = visited[current][1]
    path.reverse()
    return path
```

```python
def UC_search(G, start, goal):
    pq = [(0, start)]
    visited = {start: (0, None)}

    while pq:
        current_cost, current_node = heapq.heappop(pq)
    
        if current_node == goal:
            return current_cost, reconstruct_path(visited, start, goal)

        for neighbor, cost in G[current_node]:
            total_cost = current_cost + cost
            if neighbor not in visited or total_cost < visited[neighbor][0]:
                visited[neighbor] = (total_cost, current_node)
                heapq.heappush(pq, (total_cost, neighbor))
    
    return None

UC_search(G, "A", "M")
```

#### Iterative Deepening Depth First Search

```python
def ID_DFS(G, start, goal):
    def dls(node, depth, visited):
        if depth == 0 and node == goal:
            return [node]
        if depth > 0:
            for neighbor in G.get(node, []):
                if neighbor in visited:
                    continue
                visited.add(neighbor)
                path = dls(neighbor[0], depth - 1, visited)
                if path:
                    return [node] + path
        return None

    depth = 0
    while True:
        visited = set([start])
        path = dls(start, depth, visited)
        if path is not None:
            return path
        depth += 1
    
ID_DFS(G, 'A', 'M')
```
