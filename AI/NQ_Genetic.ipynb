# Implement N Queens Problem using Genetic Algorithm

```python
import random
```

```python
def printQ(Q):
    n = len(Q)
    for i in range(n):
        for j in range(n):
            if j+1 == Q[i]: print('Q', end = ' ')
            else: print('.', end = ' ')
        print()
```

```python
def attacks(Q, i):
    atk = 0
    for j in range(len(Q)):
        if Q[i] == Q[j]: atk += 1
        if abs(Q[i] - Q[j]) == abs(i - j): atk += 1
    return atk - 2
```

```python
def fitness(Q):
    n = len(Q)
    fitness = n * (n - 1) // 2
    for i, q in enumerate(Q):
        fitness -= attacks(Q, i)
    return fitness
```

```python
def selection(Q1, Q2):
    return Q1 if fitness(Q1) > fitness(Q2) else Q2
```

```python
def crossover(Q1, Q2):
    n = len(Q1)
    crossover_point = random.randint(1, n - 1)
    
    Qc = Q1[:crossover_point]
    
    for q in Q2[crossover_point:]:
        Qc.append(q)
    
    return Qc
```

```python
def mutate(Q, mutation_rate):
    if random.random() < mutation_rate:
        n = len(Q)
        i, j = random.sample(range(n), 2)
        Q[i], Q[j] = Q[j], Q[i]
    return Q
```

```python
def solve(n, population_size=100, max_generations=10000, mutation_rate=0.1):
    max_fitness = n * (n - 1) // 2
    population = []
    
    for _ in range(population_size):
        Q = list(range(1, n + 1))
        random.shuffle(Q)
        population.append(Q)
    
    generation = 0
    best_solution = None
    best_fitness = -float('inf')
    
    while generation < max_generations:
        fitness_scores = [(Q, fitness(Q)) for Q in population]
        fitness_scores.sort(key=lambda x: x[1], reverse=True)
        
        if fitness_scores[0][1] == max_fitness:
            best_solution = fitness_scores[0][0]
            best_fitness = fitness_scores[0][1]
            print(f"Solution found in generation {generation}!")
            break
        
        if fitness_scores[0][1] > best_fitness:
            best_fitness = fitness_scores[0][1]
            best_solution = fitness_scores[0][0][:]
        
        if generation % 100 == 0:
            print(f"Generation {generation}: Best fitness = {best_fitness}")
        
        new_population = []
        
        best_count = population_size // 10
        for i in range(best_count):
            new_population.append(fitness_scores[i][0][:])
        
        while len(new_population) < population_size:
            Q1 = selection(random.choice(population), random.choice(population))
            Q2 = selection(random.choice(population), random.choice(population))
            
            Qc = crossover(Q1, Q2)
            
            Qc = mutate(Qc, mutation_rate)
            
            new_population.append(Qc)
        
        population = new_population
        generation += 1
    
    print(f"\nFinal Results after {generation} generations:")
    print(f"Best fitness achieved: {best_fitness}")
    print(f"Best solution: {best_solution}")
    print("\nBoard configuration:")
    printQ(best_solution)
    
    return best_solution, best_fitness

if __name__ == "__main__":
    n = 69420
    population_size = 1000000000
    max_generations = 10000000000000000000
    mutation_rate = 0.1
    solve(n, population_size, max_generations, mutation_rate)
```
