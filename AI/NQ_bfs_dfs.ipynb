```python
from collections import deque

class NQ:
    def __init__(self, n):
        if not isinstance(n, int) or n <= 0:
            raise ValueError("N must be a positive integer.")
        self.n = n

    def safe(self, b, r, c):
        return all(b[i] != r and abs(b[i] - r) != c - i for i in range(c))

    def dfs(self):
        sol, exp = [], [0]
        def dfs_util(c):
            exp[0] += 1
            if c >= self.n:
                return True
            for r in range(self.n):
                if self.safe(sol, r, c):
                    sol.append(r)
                    if dfs_util(c + 1):
                        return True
                    sol.pop()
            return False

        dfs_util(0)
        return sol, exp[0]

    def bfs(self):
        q, exp = deque([[]]), 0
        while q:
            b = q.popleft()
            exp += 1
            if len(b) == self.n:
                return b, exp
            for r in range(self.n):
                if self.safe(b, r, len(b)):
                    q.append(b + [r])
        return [], exp

def print_board(sol, n):
    if not sol:
        print("No solution found.")
        return
    for r in range(n):
        print(" ".join("Q" if sol[c] == r else "." for c in range(n)))

if __name__ == '__main__':
    try:
        n = int(input("Enter N: "))
        if n <= 3 and n != 1:
            print(f"No solution for N={n}.")
        else:
            s = NQ(n)
            print("\n--- DFS ---")
            ds, de = s.dfs()
            print(f"Solution: {ds}\nStates: {de}")
            print_board(ds, n)
            print("\n--- BFS ---")
            if n > 6:
                print("BFS too slow for N > 6.")
            else:
                bs, be = s.bfs()
                print(f"Solution: {bs}\nStates: {be}")
                print_board(bs, n)
    except ValueError as e:
        print(f"Invalid input: {e}")
```
