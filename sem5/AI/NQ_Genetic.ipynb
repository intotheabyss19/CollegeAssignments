# Implement N Queens Problem using Genetic Algorithm

```python
import random
```

```python
def printQ(Q):
    n = len(Q)
    for i in range(n):
        for j in range(n):
            if j+1 == Q[i]: print('Q', end = ' ')
            else: print('.', end = ' ')
        print()
```

```python
def attacks(Q, i):
    atk = 0
    for j in range(len(Q)):
        if Q[i] == Q[j]: atk += 1
        if abs(Q[i] - Q[j]) == abs(i - j): atk += 1
    return atk - 2
```

```python
def fitness(Q):
    n = len(Q)
    fit = n * (n - 1) // 2
    for i, q in enumerate(Q):
        fit -= attacks(Q, i)
    return fit
```

```python
def selection(Q1, Q2):
    return Q1 if fitness(Q1) > fitness(Q2) else Q2
```

```python
def crossover(Q1, Q2):
    n = len(Q1)
    crosspt = random.randint(1, n - 1)
    
    Qc = Q1[:crosspt]
    
    for q in Q2[crosspt:]:
        Qc.append(q)
    
    return Qc
```

```python
def mutate(Q, mutn_rate):
    if random.random() < mutn_rate:
        n = len(Q)
        i, j = random.sample(range(n), 2)
        Q[i], Q[j] = Q[j], Q[i]
    return Q
```

```python
def solve(n, pop_size=100, max_gen=10000, mut_rate=0.1):
    max_fitness = n * (n - 1) // 2
    populn = []
    
    for _ in range(pop_size):
        Q = list(range(1, n + 1))
        random.shuffle(Q)
        populn.append(Q)
    
    gen = 0
    best_sol = None
    best_fit = -float('inf')
    
    while gen < max_gen:
        fit_scores = [(Q, fitness(Q)) for Q in populn]
        fit_scores.sort(key=lambda x: x[1], reverse=True)
        
        if fit_scores[0][1] == max_fitness:
            best_soln = fit_scores[0][0]
            best_fit = fit_scores[0][1]
            print(f"Solution found in generation {gen}!")
            break
        
        if fit_scores[0][1] > best_fit:
            best_fit = fit_scores[0][1]
            best_soln = fit_scores[0][0][:]
        
        if gen % 100 == 0:
            print(f"Generation {gen}: Best fitness = {best_fit}")
        
        new_pop = []
        
        best_count = pop_size // 10
        for i in range(best_count):
            new_pop.append(fit_scores[i][0][:])
        
        while len(new_pop) < pop_size:
            Q1 = selection(random.choice(populn), random.choice(populn))
            Q2 = selection(random.choice(populn), random.choice(populn))
            
            Qc = crossover(Q1, Q2)
            
            Qc = mutate(Qc, mut_rate)
            
            new_pop.append(Qc)
        
        populn = new_pop
        gen += 1
    
    print(f"\nFinal Results after {gen} generations:")
    print(f"Best fitness achieved: {best_fit}")
    print(f"Best solution: {best_soln}")
    print("\nBoard configuration:")
    printQ(best_soln)
    
    return best_soln, best_fit

if __name__ == "__main__":
    solve(n = 8, pop_size=100, max_gen=10000, mut_rate=0.1)
```
