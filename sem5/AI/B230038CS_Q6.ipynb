# 6. Write a program to find an optimal path for graph traversal using IDA* and recursive best first search algorithm

```python
import heapq

G = {
    'A': [('B', 3), ('C', 5), ('D', 4)],
    'B': [('E', 2), ('F', 4), ('G', 7), ('H', 4)],
    'C': [('F', 3), ('H', 2), ('I', 4), ('O', 7)],
    'D': [('J', 9)],
    'E': [('K', 2)],
    'F': [('K', 1)],
    'G': [],
    'H': [('L', 9), ('M', 7)],
    'I': [('M', 3), ('N', 2)],
    'J': [('O', 3), ('P', 1)],
    'L': [],
    'K': [],
    'M': [],
    'N': [],
    'O': [],
    'P': []
}

heuristic = {
    'A': 7, 'B': 6, 'C': 2, 'D': 1,
    'E': 5, 'F': 1, 'G': 0, 'H': 3,
    'I': 1, 'J': 4, 'K': 0, 'L': 0,
    'M': 2, 'N': 1, 'O': 0, 'P': 0
}
```

#### IDA*

```python
def IDAs(G, start, goal):
    def search(path, g, bound):
        node = path[-1]
        f = g + heuristic[node]
        if f > bound:
            return f
        if node == goal:
            return path
        min_val = float('inf')
        for neighbor, cost in G.get(node, []):
            if neighbor not in path:
                path.append(neighbor)
                t = search(path, g + cost, bound)
                if isinstance(t, list):
                    return t
                if t < min_val:
                    min_val = t
                path.pop()
        return min_val

    bound = heuristic[start]
    path = [start]
    while True:
        t = search(path, 0, bound)
        if isinstance(t, list):
            return t
        if t == float('inf'):
            return None
        bound = t
IDAs(G, 'A', 'M')
```

#### Recursive Best First Search

```python
def RBFS(G, start, goal):
    def rbfs_recursive(node, f_limit, g_cost, path):
        if node == goal:
            return path, 0
        
        successors = []
        for neighbor, move_cost in G.get(node, []):
            g = g_cost + move_cost
            h = heuristic[neighbor]
            f = max(g + h, g_cost + heuristic[node])
            successors.append((f, neighbor))
            
        if not successors:
            return None, float('inf')

        while True:
            successors.sort()
            best_f, best_node = successors[0]
            
            if best_f > f_limit:
                return None, best_f
            
            alternative_f = successors[1][0] if len(successors) > 1 else float('inf')
            
            new_path, backed_up_f = rbfs_recursive(best_node, min(f_limit, alternative_f), g_cost + G[node][0][1], path + [best_node])

            if new_path is not None:
                return new_path, backed_up_f

            successors[0] = (backed_up_f, best_node)

    path, cost = rbfs_recursive(start, float('inf'), 0, [start])
    return path

RBFS(G, 'A', 'M')
```
