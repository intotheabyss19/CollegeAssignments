```python
puzzle.ipynbpuzzle.ipynb
import random
import os
import time

Grid = [7, 2, 4, 5, 0, 6, 8, 3, 1]
Goal = [0, 1, 2, 3, 4, 5, 6, 7, 8]

def printG(grid):
    """Print 3x3 grid properly"""
    for i in range(0, 9, 3):
        print(grid[i:i+3])
    print()

def clear_screen():
    os.system('clear')

class Puzzle:
    def __init__(self, grid, goal):
        self.grid = grid[:]  # Create a copy to avoid modifying original
        self.goal = goal
    
    def find_index(self, num):
        """Find index of a number in the grid"""
        for i, val in enumerate(self.grid):
            if val == num:
                return i
        return -1  # Not found
    
    def get_valid_moves(self):
        """Get valid moves for the empty space (0)"""
        empty_pos = self.find_index(0)
        valid_moves = [
            (1, 3),           # Position 0
            (0, 2, 4),        # Position 1
            (1, 5),           # Position 2
            (0, 4, 6),        # Position 3
            (1, 3, 5, 7),     # Position 4
            (2, 4, 8),        # Position 5
            (3, 7),           # Position 6
            (4, 6, 8),        # Position 7
            (5, 7)            # Position 8
        ]
        return valid_moves[empty_pos]
    
    def move(self, target_pos):
        """Move a tile to the empty space"""
        empty_pos = self.find_index(0)
        if target_pos in self.get_valid_moves():
            # Swap empty space with target position
            self.grid[empty_pos], self.grid[target_pos] = self.grid[target_pos], self.grid[empty_pos]
            return True
        return False
    
    def is_solved(self):
        return self.grid == self.goal

def ReflexAgent():
    puzzle = Puzzle(Grid, Goal)
    moves_count = 0
    max_moves = 1000  # Prevent infinite loops
    
    print("Initial state:")
    printG(puzzle.grid)
    
    while not puzzle.is_solved() and moves_count < max_moves:
        valid_moves = puzzle.get_valid_moves()
        
        # Find moves where tile is NOT in correct position
        non_correct_moves = []
        correct_moves = []
        
        for move_pos in valid_moves:
            tile_value = puzzle.grid[move_pos]
            if tile_value == move_pos:  # Tile is in correct position
                correct_moves.append(move_pos)
            else:
                non_correct_moves.append(move_pos)
        
        # Reflex agent logic: avoid moving correctly placed tiles 80% of time
        if correct_moves and non_correct_moves and random.random() < 0.8:
            # 80% chance: choose from non-correct positions
            chosen_move = random.choice(non_correct_moves)
        else:
            # 20% chance: choose any valid move (including correct ones)
            chosen_move = random.choice(valid_moves)
        
        # Make the move
        puzzle.move(chosen_move)
        moves_count += 1
        
        # Display current state
        clear_screen()
        print(f"Move {moves_count}:")
        printG(puzzle.grid)
        time.sleep(0.5)  # Slow down for visualization
    
    if puzzle.is_solved():
        print(f"Solved in {moves_count} moves!")
    else:
        print(f"Max moves ({max_moves}) reached without solution")

# Run the agent
ReflexAgent()

Grid = [7, 2, 4, 5, 0, 6, 8, 3, 1]
Goal = [0, 1, 2, 3, 4, 5, 6, 7, 8]
```
