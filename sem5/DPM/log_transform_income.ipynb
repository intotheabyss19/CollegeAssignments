# Applying Logarithmic Transformation to Reduce Skewness


This notebook demonstrates how to:
- Apply a **logarithmic transformation** to a numerical feature (`income`)
- Reduce **skewness** and normalize the data distribution
    

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Sample dataset
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'income': [25000, 30000, 45000, 120000, 250000, 1000000]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
display(df)

plt.figure()
df['income'].hist(bins=10)
plt.title("Original Income Distribution")
plt.show()
```

```python

def log_transform(df):
    # Apply logarithmic transformation to 'income'
    df['log_income'] = np.log1p(df['income'])  # log1p handles zero safely (log(1+x))
    return df

df_transformed = log_transform(df)
print("Transformed DataFrame:")
display(df_transformed)

plt.figure()
df_transformed['log_income'].hist(bins=10)
plt.title("Log-Transformed Income Distribution")
plt.show()
```


### ✅ Explanation
- The `np.log1p()` function is used instead of `np.log()` to handle possible zero incomes.
- The transformation compresses large income values, reducing **right-skewness**.
- The resulting histogram shows a more **normalized** distribution.
