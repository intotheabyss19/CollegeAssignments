# Decision Tree using Hunt’s Algorithm (Borrower Dataset)


This notebook demonstrates how to **design a decision tree** using the **general structure of Hunt’s algorithm**
on the given *Borrower dataset*.

### Hunt’s Algorithm (General Structure)
1. **If all records** belong to the same class → make that a **leaf node**.  
2. **If records are mixed** → choose the best attribute to split (using entropy or Gini).  
3. **Partition** the data according to the selected attribute’s values.  
4. **Recursively apply** the same procedure to each subset until all leaf nodes are pure.
    

```

import pandas as pd

# Borrower dataset
data = {
    'Home Owner': ['Yes','No','No','Yes','No','No','Yes','No','No','No'],
    'Marital Status': ['Single','Married','Single','Married','Divorced','Married','Divorced','Single','Married','Single'],
    'Annual Income': [125,100,70,120,95,60,220,85,75,90],
    'Defaulted Borrower': ['No','No','No','No','Yes','Yes','No','Yes','No','Yes']
}

df = pd.DataFrame(data)
df
```

```

from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Prepare data
X = df[['Home Owner', 'Marital Status', 'Annual Income']]
y = df['Defaulted Borrower']

# Convert categorical variables to dummy variables
X = pd.get_dummies(X, drop_first=True)

# Train Decision Tree
model = DecisionTreeClassifier(criterion='entropy', random_state=0)
model.fit(X, y)

# Plot the tree
plt.figure(figsize=(12,7))
plot_tree(model, feature_names=X.columns, class_names=['No','Yes'], filled=True, rounded=True)
plt.title("Decision Tree (Generated using Hunt's Algorithm Structure)")
plt.show()
```


### ✅ Explanation
- **Hunt’s algorithm** forms the conceptual basis for decision tree learning methods such as ID3, C4.5, and CART.
- The model splits based on **entropy** to maximize information gain.
- Each leaf node represents a **class label** (“Yes” or “No” for default).
- The resulting tree visually demonstrates the step-by-step partitioning of the dataset.
