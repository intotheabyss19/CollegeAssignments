# Discretizing 'age' and Binarizing 'gender' in Python


This notebook demonstrates how to:
1. **Discretize** a continuous numerical feature (`age`) into categorical bins:  
   - 0–18 → 'Child'  
   - 19–35 → 'Young Adult'  
   - 36–60 → 'Adult'  
   - 61+ → 'Senior'
2. **Binarize** a categorical feature (`gender`) into binary columns using one-hot encoding.
    

```python

import pandas as pd

data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'age': [15, 23, 42, 67, 30, 80],
    'gender': ['Female', 'Male', 'Male', 'Male', 'Female', 'Female']
}

df = pd.DataFrame(data)
print("Original DataFrame:")
display(df)
```

```python
def preprocess_data(df):
    bins = [0, 18, 35, 60, float('inf')]
    labels = ['0-18', '19-35', '36-60', '61+']
    df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)
    
    df = pd.get_dummies(df, columns=['gender'], prefix='gender')
    
    return df

processed_df = preprocess_data(df)
print("Processed DataFrame:")
display(processed_df)
```


### ✅ Output Explanation
- The **age** column is converted into `age_group` with discrete ranges.
- The **gender** column is converted into binary columns (`gender_Female`, `gender_Male`).
