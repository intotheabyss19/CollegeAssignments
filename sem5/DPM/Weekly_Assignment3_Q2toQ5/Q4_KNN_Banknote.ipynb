# Q4 — KNN on Banknote Authentication Dataset

Loads the dataset from UCI (if online) and runs a K-Nearest Neighbors classifier with basic train/test split and evaluation.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import os

# UCI dataset URL (raw data)
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00267/data_banknote_authentication.txt'

# Try to load dataset; if not available (no internet), create a small synthetic sample
try:
    df = pd.read_csv(url, header=None)
    df.columns = ['variance','skewness','curtosis','entropy','class']
    print('Loaded dataset from UCI, shape =', df.shape)
except Exception as e:
    print('Could not load from URL (no internet?). Using a small synthetic sample. Error:', e)
    data = {
        'variance': [2.3, 0.8, -2.1, 1.5, -1.1],
        'skewness': [6.0, 3.2, -4.2, 2.1, -2.0],
        'curtosis': [0.5, 1.2, -2.3, 0.1, -1.1],
        'entropy': [0.0, -1.2, 2.3, -0.6, 1.0],
        'class': [0,0,1,0,1]
    }
    df = pd.DataFrame(data)

df.head()
```

```python
# Prepare data
X = df.drop(columns=['class'])
y = df['class']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

# Fit KNN with k=3
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nClassification report:\n', classification_report(y_test, y_pred))
print('\nConfusion matrix:\n', confusion_matrix(y_test, y_pred))
```
